```python
# For EuroCup vs EuroLeague
ec_summary = utils.compare_league_correlations(eurocup_vs_euroleague, 'EC', 'EL')

# For Liga A vs EuroLeague
la_summary = utils.compare_league_correlations(ligaa_vs_euroleague, 'LA', 'EL')

# For ACB vs EuroLeague
acb_summary = utils.compare_league_correlations(acb_vs_euroleague, 'ACB', 'EL')
```

In interpreting the following summaries, a high correlation means that that metric is highly related from the lower league to the higher league (EuroLeague). The average ratio from EuroLeague to lower league is interpreted as the factor in which that metric changes from the lower league to the EuroLeague on average. The exception to this is internal box plus minus - since this is not a stat that is always positive, the change metric for this is simply the difference (EL stat number - EC stat number).

For example, a ratio of 1.09 for total rebounding percentage with a correlation of 0.77 in the EuroCup data means that there is a pretty high correlation between the total rebounding percentage achieved in the EC compared to the total rebounding percentage achieved in the EL, and that on average the rebounding percentage in the EL is actually 9% higher.

**Takeaways:**
- Assist percentage has a high correlation in all three leagues to the EL. Typically assist percentage goes up when transferring to the EL, except for the EuroCup.
- Total rebounding percentage has correlations of at least 0.5 in all three leagues as well, with the ratio positive in all three (higher % when playing in the EL).
- Three point attempt rate has a high correlation in all three leagues to the EL. The rate goes up in the EL in all three instances.
- Usage percentage has a ratio of about 1 for all leagues, although correlation varies.
- **True shooting percentage has a correlation under 0.5 in all three instances but the ratio intuitively makes sense as it is below 1 in all three. This combined with the rebounding percentage ratio being positive in all three indicates that the defense is at a higher level in the EuroLeague. As we explore upcoming, this is good to know that strong true shooting and offensive performance is most impressive if done in the EuroLeague.**
- The internal box plus minus is the most important stat but it has low correlation. That being said, the change metric is negative in all three instances, showing that the EuroLeague is the best competition and players are less effective playing there.

```python
leagues = {
    'EuroLeague': agg_el_df,
    'EuroCup': agg_ec_df,
    'Spain - ACB': agg_acb_df,
    'Italy - Liga A': agg_la_df
}
```

Since the goal is to analyze 2021 international players - it is obvious that none of those players would have made the NBA yet. So for analysis purposes of figuring out what leads to a player making the jump to the NBA, we will want to exclude them.

```python
agg_el_df_without_2021 = agg_el_df[agg_el_df['latest_season_before_nba'] < 2021]
agg_ec_df_without_2021 = agg_ec_df[agg_ec_df['latest_season_before_nba'] < 2021]
agg_acb_df_without_2021 = agg_acb_df[agg_acb_df['latest_season_before_nba'] < 2021]
agg_la_df_without_2021 = agg_la_df[agg_la_df['latest_season_before_nba'] < 2021]
```

```python
leagues_without_2021 = {
    'EuroLeague': agg_el_df_without_2021,
    'EuroCup': agg_ec_df_without_2021,
    'Spain - ACB': agg_acb_df_without_2021,
    'Italy - Liga A': agg_la_df_without_2021
}
```

```python
# Calculate the percentage for each league
data = []
for league_name, df in leagues_without_2021.items():
    pct = (df['jump'].sum() / len(df)) * 100
    data.append({'League': league_name, 'Percentage of Players that Made the NBA': pct})

plot_df = pd.DataFrame(data)

# Plot
plt.figure(figsize=(8, 5))
sns.barplot(
    data=plot_df,
    x='League',
    y='Percentage of Players that Made the NBA',
    palette='Blues_d'
)

plt.title('Percentage of Players that Made the NBA by League (not including 2021)', fontsize=14)
plt.ylabel('Percentage (%)')
plt.xlabel('')
plt.ylim(0, 100)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
for league_name, df in leagues.items():
    # Count players whose latest international season is 2021 and have not played in the NBA
    count = len(df[
        (df['latest_season_before_nba'] == 2021) &
        (df['earliest_nba_season'].isna())
    ])
    
    print(f"{count} players in {league_name} are still playing internationally in 2021 "
          f"that have not played in the NBA that we can analyze.")
```

Applying a function to see what percentage of players that made the NBA from lower leagues also played in the EuroLeague. Since the EL is the highest competition, it would make sense a high number of players that made the NBA from lower leagues also played in the EuroLeague.

```python
# Apply to lower leagues
utils.league_to_el_overlap(agg_ec_df, "EuroCup", agg_el_df)
utils.league_to_el_overlap(agg_acb_df, "Spain - ACB", agg_el_df)
utils.league_to_el_overlap(agg_la_df, "Italy - Liga A", agg_el_df)
```

We already saw that the EuroLeague has a higher percentage of players that make it to the NBA, but that competition disparity goes even further. Of the next two leagues with the most players sent to the NBA, more than half of those players also played in the EuroLeague. This shows us that we should be spending most of our efforts looking at the EuroLeague.

```python
avg_ibpm_data = []

for league_name, df in leagues.items():
    # Filter players who made the jump
    jump_df = df[df['jump'] == True]
    
    # Compute average NBA internal box plus minus (if column exists)
    if not jump_df.empty and 'avg_nba_ibpm' in jump_df.columns:
        avg_ibpm = jump_df['avg_nba_ibpm'].mean()
    else:
        avg_ibpm = float('nan')
    
    avg_ibpm_data.append({
        'League': league_name,
        'Average NBA iBPM': avg_ibpm
    })

# Create a dataframe for plotting
avg_ibpm_df = pd.DataFrame(avg_ibpm_data)
```

```python
# Plot
plt.figure(figsize=(8, 5))
plt.bar(avg_ibpm_df['League'], avg_ibpm_df['Average NBA iBPM'], color='skyblue', edgecolor='black')

# Add titles and labels
plt.title("Average NBA Internal Box Plus Minus (iBPM)\nfor Players Who Made the Jump", fontsize=14, weight='bold')
plt.ylabel("Average NBA iBPM", fontsize=12)
plt.xlabel("League", fontsize=12)
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add values above bars
for i, val in enumerate(avg_ibpm_df['Average NBA iBPM']):
    plt.text(i, val + 0.1, f"{val:.2f}", ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

**Additionally, for players that do make the NBA - the players coming from the EuroLeague typically are more effective as well.**

### Feature Analysis

Applying a function that runs LASSO logistic regression to predict 'jump', selects significant features,
and refits a statsmodels Logit model for p-values.

```python
model_el, summary_el = utils.run_lasso_logistic_with_pvalues(agg_el_df_without_2021, "EuroLeague")
```

**Takeaways:**
- Overall, the model classifies players correctly ~87% of the time
- Younger players are much more likely to jump to the NBA — older players have sharply lower odds. Age is the only significant feature.
- Playing more minutes per game increases odds of making the NBA — suggests playing a big role on your team matters.
- ORB % has negative coefficient - Players who crash the offensive boards more are less likely to make the NBA — possibly because that profile fits bigs who may lack perimeter versatility.
- Higher overall impact on court (by this internal box plus minus) increases odds of NBA jump.

```python
model_ec, summary_ec = utils.run_lasso_logistic_with_pvalues(agg_ec_df_without_2021, "EuroCup")
```

Here in the EuroCup, nothing after age has a statistical significance. However, both internal box plus minus and free throw rate have p values under 0.1 and have positive coefficients. The correlation of internal box plus minus is obvious, but the free throw rate's significance could be because free throw rate is typically higher for better athletes/drivers. Without data like height, weight, and combine metrics, I can't measure this but it is a good indicator.

```python
model_acb, summary_acb = utils.run_lasso_logistic_with_pvalues(agg_acb_df_without_2021, "Spain - ACB")
```

Age, minutes, internal box plus minus, and steal percentage all have p values under 0.05. True shooting percentage and turnover percentage also have p values under 0.1. Steal percentage is something that a lot of analysts look at when analyzing college athletes as well because it usually has some correlation to basketball IQ.

```python
model_la, summary_la = utils.run_lasso_logistic_with_pvalues(agg_la_df_without_2021, "Italy - Liga A")
```

Only age, defensive rebounding percentage, and block percentage have p values under 0.1. After analyzing all of these models, it's clear that age and internal box plus minus are the most consistent indicators of success translating to the NBA. If I had more time, I could run more models analyzing what features predicted players jumping from one of these lower leagues to the EuroLeague, so that we could identify players to keep an eye on but necessarily bring over right away. For the purpose of finding players to scout for this year, this is not necessary at this time.

The biggest features across all are **age, minutes, and internal box plus minus**, and for some secondary metrics, **free throw rate, steal percentage, and true shooting percentage could also be helpful.**

1. The non-negotiables for the players transferring to the NBA.

- more than 300 minutes

- free throw rate over 20%
